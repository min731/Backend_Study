```python
from dotenv import load_dotenv
load_dotenv()

from langgraph.prebuilt import create_react_agent

def get_weather(city: str) -> str:  
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_react_agent(
    model="gpt-4o-mini",  
    tools=[get_weather],  
    prompt="You are a helpful assistant"  
)

# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]}
)
```

```python
from langchain_core.messages import AnyMessage
from langchain_core.runnables import RunnableConfig
from langgraph.prebuilt.chat_agent_executor import AgentState
from langgraph.prebuilt import create_react_agent

def prompt(state: AgentState, config: RunnableConfig) -> list[AnyMessage]:  
    user_name = config["configurable"].get("user_name")
    system_msg = f"You are a helpful assistant. Address the user as {user_name}."
    return [{"role": "system", "content": system_msg}] + state["messages"]

agent = create_react_agent(
    model="gpt-4o-mini",
    tools=[get_weather],
    prompt=prompt
)

agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]},
    config={"configurable": {"user_name": "John Smith"}}
)
```

```python
from langgraph.prebuilt import create_react_agent
from langgraph.checkpoint.memory import InMemorySaver

checkpointer = InMemorySaver()

agent = create_react_agent(
    model="gpt-4o-mini",
    tools=[get_weather],
    checkpointer=checkpointer  
)

# Run the agent
config = {"configurable": {"thread_id": "1"}}
sf_response = agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]},
    config  
)
ny_response = agent.invoke(
    {"messages": [{"role": "user", "content": "what about new york?"}]},
    config
)
```

```python
from langgraph.graph import StateGraph, END
from typing import TypedDict, Annotated, List
import operator

# 상태 정의
class DocumentProcessingState(TypedDict):
    messages: Annotated[List[str], operator.add]
    document_text: str
    processed_chunks: List[str] 
    embeddings: List[List[float]]
    search_query: str
    search_results: List[str]
    current_step: str

# Node 함수들 정의 (각각이 처리 단위)
def document_loader_node(state: DocumentProcessingState) -> DocumentProcessingState:
    """문서 로딩 노드"""
    doc_text = "이것은 머신러닝에 관한 문서입니다. 딥러닝은 인공신경망을 사용합니다."
    
    return {
        **state,
        "messages": state["messages"] + ["문서 로딩 완료"],
        "document_text": doc_text,
        "current_step": "document_loaded"
    }

def text_splitter_node(state: DocumentProcessingState) -> DocumentProcessingState:
    """텍스트 분할 노드"""
    doc_text = state["document_text"]
    # 간단한 문장 분할
    chunks = [sentence.strip() for sentence in doc_text.split('.') if sentence.strip()]
    
    return {
        **state,
        "messages": state["messages"] + [f"{len(chunks)}개 청크로 분할 완료"],
        "processed_chunks": chunks,
        "current_step": "text_split"
    }

def embedding_node(state: DocumentProcessingState) -> DocumentProcessingState:
    """임베딩 생성 노드"""
    chunks = state["processed_chunks"]
    # 가짜 임베딩 생성 (실제로는 모델 사용)
    embeddings = [[0.1, 0.2, 0.3] for _ in chunks]
    
    return {
        **state,
        "messages": state["messages"] + ["임베딩 생성 완료"],
        "embeddings": embeddings,
        "current_step": "embeddings_created"
    }

def search_node(state: DocumentProcessingState) -> DocumentProcessingState:
    """검색 노드"""
    query = state["search_query"]
    chunks = state["processed_chunks"]
    
    # 간단한 키워드 검색 시뮬레이션
    results = [chunk for chunk in chunks if any(word in chunk.lower() for word in query.lower().split())]
    
    return {
        **state,
        "messages": state["messages"] + [f"검색 완료: {len(results)}개 결과"],
        "search_results": results,
        "current_step": "search_completed"
    }

# Edge 조건 함수 (흐름 제어)
def decide_next_step(state: DocumentProcessingState) -> str:
    """다음 단계 결정"""
    current = state["current_step"]
    
    if current == "document_loaded":
        return "text_splitter"
    elif current == "text_split":
        return "embedding_generator" 
    elif current == "embeddings_created":
        if state.get("search_query"):  # 검색 쿼리가 있으면
            return "searcher"
        else:
            return END
    elif current == "search_completed":
        return END
    else:
        return END

# LangGraph 구성
def create_document_processing_graph():
    """문서 처리 그래프 생성"""
    
    workflow = StateGraph(DocumentProcessingState)
    
    # Node 추가 (각 처리 단위)
    workflow.add_node("document_loader", document_loader_node)
    workflow.add_node("text_splitter", text_splitter_node) 
    workflow.add_node("embedding_generator", embedding_node)
    workflow.add_node("searcher", search_node)
    
    # Edge 추가 (노드 간 연결)
    workflow.set_entry_point("document_loader")  # 시작점
    
    # 조건부 엣지 (상태에 따른 분기)
    workflow.add_conditional_edges(
        "document_loader",
        decide_next_step,
        {
            "text_splitter": "text_splitter"
        }
    )
    
    workflow.add_conditional_edges(
        "text_splitter", 
        decide_next_step,
        {
            "embedding_generator": "embedding_generator"
        }
    )
    
    workflow.add_conditional_edges(
        "embedding_generator",
        decide_next_step,
        {
            "searcher": "searcher",
            END: END
        }
    )
    
    workflow.add_edge("searcher", END)  # 직접 엣지
    
    return workflow.compile()

# 사용 예시
print("=== LangGraph 예제 ===")
graph_app = create_document_processing_graph()

# 검색 쿼리 없이 실행
result1 = graph_app.invoke({
    "messages": [],
    "search_query": "",
    "current_step": "start"
})

print("검색 없는 경우:")
for msg in result1["messages"]:
    print(f"  - {msg}")

# 검색 쿼리와 함께 실행  
result2 = graph_app.invoke({
    "messages": [],
    "search_query": "머신러닝",
    "current_step": "start"
})

print("\n검색 있는 경우:")
for msg in result2["messages"]:
    print(f"  - {msg}")
print(f"검색 결과: {result2['search_results']}")
```
